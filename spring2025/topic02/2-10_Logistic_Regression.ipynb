```python
import matplotlib.pyplot as plt
import numpy as np

# Define the sigmoid function
def sigmoid(x, a=1, b=0):
    return 1 / (1 + np.exp(-(a*x + b)))

# Set the range of x values
x_values = np.linspace(-5, 5, 100)

# Compute the sigmoid function values for b = 0 and b = 2
sigmoid_b0_values = sigmoid(x_values, a=1, b=0)
sigmoid_b2_values = sigmoid(x_values, a=1, b=2)

# Plot the graph
plt.figure(figsize=(8, 5))
plt.plot(x_values, sigmoid_b0_values, label="b = 0", linestyle='-')
plt.plot(x_values, sigmoid_b2_values, label="b = 2", linestyle='--')

# Customize the graph
plt.xlabel("x")
plt.ylabel("Sigmoid(x + b)")
plt.title("Sigmoid Function with Different b Values")
plt.legend()
plt.grid()

# Display the graph
plt.show()
```

# **Logistic Regression**

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
w = np.arange(-10, 10, 0.001)
b = 1
x = np.array([1.3, 1.2, 3.5, 4.1])
y = np.array([0, 0, 1, 1])
cost = []
for i in range(len(w)):
    cost.append(sum((w[i] * x + b - y) ** 2) / len(x))
```

```python
plt.plot(w, cost, 'ro', markersize=0.1)
plt.axis([-12, 15, -100, 1000])
plt.xlabel("w")
plt.ylabel("Cost(w)")
plt.show()
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
cost = []
for i in range(len(w)):
    cost.append(sum((sigmoid(w[i] * x + b) - y) ** 2) / len(x))
```

```python
plt.plot(w, cost, 'ro', markersize=0.1)
plt.axis([-7.5, 7.5, 0.1, 0.7])
plt.xlabel("w")
plt.ylabel("Cost(w)")
plt.show()
```

```python
cost = []
for i in range(len(w)):
    cost.append(sum(-y*np.log(sigmoid(w[i]*x+b))-(1-y)*np.log(1-sigmoid(w[i]*x+b))))
```

```python
plt.plot(w, cost, 'ro', markersize=0.1)
plt.xlabel("w")
plt.ylabel("Cost(w)")
plt.show()
```
