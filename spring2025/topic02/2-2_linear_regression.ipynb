# **Linear Regression**

```python
import numpy as np
#a_and_b = np.array([3.4, 77]) # a and b
a = 3.4
b = 77
data = np.array([[2, 81], [4, 93], [6, 91], [8, 97]])
train_x = data[:, 0]
train_y = data[:, 1]
```

```python
print(data)
```

```python
print(train_x)
print(train_y)
```

```python
print(a * train_x + b)
```

```python
mse = sum(((a * train_x + b) - train_y)**2)/len(train_x)
print(mse)
```
