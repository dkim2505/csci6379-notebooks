```python
import numpy as np

def softmax(x, axis=-1):
    # Subtract the maximum value for numerical stability
    #x_exp = np.exp(x - np.max(x, axis=axis, keepdims=True))
    x_exp = np.exp(x)
    return x_exp / np.sum(x_exp, axis=axis, keepdims=True)

def scaled_dot_product_attention(X, W_q, W_k, W_v):
    # Compute Query, Key, and Value matrices
    Q = np.dot(X, W_q)
    K = np.dot(X, W_k)
    V = np.dot(X, W_v)

    print("Q", Q)
    print("K", K)
    print("V", V)

    # Get the dimension of the key vectors (d_k)
    d_k = Q.shape[-1]

    # Compute the dot products between Query and Key, then scale by sqrt(d_k)
    scores = np.dot(Q, K.T) / np.sqrt(d_k)

    print("scores", scores)

    # Apply softmax to get the attention weights
    attn_weights = softmax(scores, axis=-1)

    print("attn_weights", attn_weights)

    # Compute the output as a weighted sum of the value vectors
    output = np.dot(attn_weights, V)
    return output

# Example values:
X = np.array([[1, 0],
              [2, 1]])

W_q = np.array([[0.1, 0.2],
                [0.3, 0.4]])

W_k = np.array([[0.5, 0.6],
                [0.7, 0.8]])

W_v = np.array([[0.9, 1.0],
                [1.1, 1.2]])

# Compute attention output and weights
output = scaled_dot_product_attention(X, W_q, W_k, W_v)

print("Output:\n", output)

```
