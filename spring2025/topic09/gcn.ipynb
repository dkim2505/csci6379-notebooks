```python
import torch
from torch_geometric.data import Data
from torch_geometric.nn import GCNConv
import torch.nn.functional as F

# Define molecular graph data: example with 4 atoms and 3 bonds
edge_index = torch.tensor([[0, 1, 1, 2, 2, 3], [1, 0, 2, 1, 3, 2]], dtype=torch.long)
x = torch.tensor([[6], [7], [8], [16]], dtype=torch.float)  # Example atom features

# Create molecular graph data
data = Data(x=x, edge_index=edge_index)

# Define a simple GNN model
class MolecularGNN(torch.nn.Module):
   def __init__(self):
       super(MolecularGNN, self).__init__()
       self.conv1 = GCNConv(1, 4)  # Input features 1 -> output features 4
       self.conv2 = GCNConv(4, 2)  # Output features 2 for final embedding

   def forward(self, data):
       x, edge_index = data.x, data.edge_index
       x = self.conv1(x, edge_index)
       x = F.relu(x)
       x = self.conv2(x, edge_index)
       return x

# Initialize model and compute output
model = MolecularGNN()
output = model(data)
print(output)  # Final embeddings for each atom (node)
```
