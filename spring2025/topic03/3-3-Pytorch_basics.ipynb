# **Basics of PyTorch**

Tensors are the heart and soul of PyTorch! To get the best out of PyTorch, you've gotta be buddies with tensors. So, why wait? Let's dive into the wonderful world of tensors together!

# **Tensor**

If you're gearing up to dive into PyTorch, it's super important to get comfy with Tensors. So, let's get our hands dirty and start with creating our very first Tensor.


`torch.tensor() and tensor.numpy()`

Think of it as PyTorch's version of NumPy's ndarray, but with a cool twist: it can be crunched on a GPU. Now, to bring a Tensor to life, here's a handy-dandy piece of code you can use.

```python
import torch
print(torch.__version__)
print(torch.tensor([[1, 2], [3, 4]])) # Create a 2D tensor
print(torch.tensor([[1, 2], [3, 4]], device='cuda:0')) # Generating a tensor on the GPU
print(torch.tensor([[1, 2], [3, 4]], dtype=torch.float64)) # Create a tensor using dtype
```

Let's convert a tensor to ndarray .

```python
temp = torch.tensor([[1, 2],[3, 4]])
print(temp.numpy()) # Convert tensor to ndarray
temp = torch.tensor([[1, 2],[3, 4]], device='cuda:0')
print(temp.to('cpu').numpy()) # Convert tensor on GPU to tensor on CPU, and then convert to ndarray
```

# **Tensor Types and Manipulations**

There are three types of tensors:

* `torch.FloatTensor`: 32-bit float

* `torch.DoubleTensor`: 64-bit float

* `torch.LongTensor`: 64-bit integer


Here's how to handle tensor indexes and slices:

```python
t = torch.FloatTensor([1, 2, 3, 4, 5, 6, 7])
print(t[0], t[1], t[-1])
print(t[2:5], t[4:])
```

# **Operations and Reshape**

Tensors, like NumPy's ndarray, allow diverse math operations. They perform faster with GPUs. Be aware, differing tensor types can cause errors.



```python
t1 = torch.tensor([1, 2, 3])
t2 = torch.tensor([4, 5, 6])
t3 = t1 + t2
print(t3)
```

To change a tensor's dimension, use **`view`**. For merging, use `stack` and **`cat`**. **`t`** and `transpose` are for swapping dimensions. **`view`** is like NumPy's `reshape`.

```python
# reshaping tensor
tensor = torch.arange(10)
print(tensor)
reshaped = tensor.view(2, 5)
print(reshaped)
print('\n')

# stacking tensors
tensor1 = torch.Tensor([1,2,3])
tensor2 = torch.Tensor([4,5,6])
stacked = torch.stack((tensor1, tensor2))
print(stacked)
print('\n')

# concatenating tensors
cat = torch.cat((tensor1, tensor2))
print(cat)
print('\n')

# transposing tensor
tensor = torch.Tensor([[1,2,3],[4,5,6]])
transposed = tensor.t()
print(transposed)
```
