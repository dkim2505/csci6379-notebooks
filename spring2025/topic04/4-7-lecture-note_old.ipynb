# **Convolutional Neural Networks**

In deep learning, to train a neural network, one must repeat the processes of forward and backward propagation, requiring the learning of all parameters, or a vast number of them. This process is complex and consumes significant computing resources. In contrast, Convolutional Neural Networks (CNNs) save time and resources by computing only local regions, enabling the efficient identification and learning of more detailed patterns in images. Let's explore this in more detail.

**Why Use CNNs When We Already Have NNs?**

CNNs are primarily used when working with image or video data (a sequence of images). Consider a simple 3x3 image. If we were to flatten this image into a vector (9x1) and use a traditional Neural Network (NN) to process it, we would lose the spatial features inherent to the image. The spatial arrangement of pixels in images carries significant information about the structure and objects within the image, which is crucial for tasks like image recognition, object detection, and many others. Traditional NNs, without a mechanism to recognize these spatial hierarchies, treat input data as a collection of independent features, losing the contextual relationships that are vital for understanding images.

CNNs, on the other hand, are designed to capture these spatial hierarchies by applying filters (kernels) that move across the image in a sliding window fashion, effectively recognizing patterns such as edges, textures, and other features in localized areas. This allows CNNs to extract and learn features from specific parts of an image, making them exceptionally well-suited for tasks involving visual data. By preserving the spatial structure of the image through the use of convolutional layers and pooling layers, CNNs can achieve higher accuracy in tasks such as image classification, object detection, and more, compared to traditional NNs that do not consider the spatial structure of the input data.

**Structure of CNNs**

The basic structure of a Convolutional Neural Network (CNN) consists of the following five layers:

* **Input Layer:** This layer takes the raw pixel values of the image as input.

* **Convolutional Layer:** This layer applies various filters to the input to create a feature map, highlighting features like edges, textures, etc.

* **Pooling Layer:** Following the convolutional layer, this layer reduces the spatial size (width, height) of the input volume for the next convolutional layer. It helps in reducing the computation required and also helps in making the detection of features invariant to scale and orientation changes.

* **Fully Connected Layer:** After several convolutional and pooling layers, the high-level reasoning in the neural network is done through fully connected layers. Neurons in a fully connected layer have full connections to all activations in the previous layer, as seen in regular Neural Networks. Their role is to take these high-level features and transform them into a final output prediction.

* **Output Layer:** This layer produces the final predictions of the network, such as the class scores in a classification task.

In a CNN, convolutional and pooling layers are typically stacked repeatedly before passing the result to the fully connected layers. This repetition allows the network to learn features at various levels of abstraction. For example, initial layers might learn to detect edges, colors, and textures, while deeper layers can learn to detect more complex patterns based on the features detected by earlier layers.

**Input Layer**

The input layer is the layer where images are fed into the network. Images are three-dimensional data consisting of height, width, and channel values. If the image is in grayscale, the channel value is 1, indicating only intensity information. For color images, the channel value is typically 3, corresponding to the three primary color channels red, green, and blue (RGB).

**Convolutional Layer**

In this layer, features from the image are extracted. Feature extraction uses kernels (or filters, terms used interchangeably) to detect features. These kernels, in the form of matrices, sweep across all areas of the input image to extract features. The output obtained from this process is referred to as a feature map. Typically, kernels are applied in sizes of 3x3 or 5x5, and the stride, which determines how much the kernel moves across the image, is commonly set to a default value of 1. In the learning process of a CNN, the values that are actually learned are these kernels. Extracting features using a kernel means performing a weighted sum of the kernel and the part of the image it is applied to, and this weighted sum produces a single pixel in the feature map.

It's important to remember that the number of feature maps generated is equal to the number of kernels used. For example, if there are 32 kernels, then 32 feature maps are produced. Additionally, if the input image has $k$ channels, then the kernels will also be composed of $k$ channels to accommodate each channel of the input. This means that for an input image with 3 channels, such as a color image using RGB (Red, Green, Blue), each kernel will have 3 channels. This structure allows the network to process and extract features from all channels of the input simultaneously. For instance, in the case with two kernels, $w_0$ and $w_1$, each would have 3 channels corresponding to the RGB channels of the input image, allowing the network to learn complex features that take into account the full color information of the input.

**Structure of CNN**

Given the input and the hyperparameters, the size of the output data from a convolutional layer can be calculated as follows:

Input Size: $w×h×d$ (width, height, depth/channel)

Hyperparameters:

* k: Number of filters (determines the depth of the output)

* f: Filter size

* s: Stride

* p: Padding

Output Size:

* Output Width: $\frac{w-f+2p}{s}+1$

* Output Height: $\frac{h-f+2p}{s}+1$

* Output Depth: $k$


**Padding**

Padding in PyTorch refers to the technique of adding extra elements or values to the input data or feature maps, usually around the borders, to preserve the spatial dimensions during convolutional operations. Padding is commonly used to ensure that the output size matches the input size or to prevent information loss at the edges of an image or feature map.

**Pooling layers**

Pooling layers reduce the dimensionality of feature maps by downsampling, thereby decreasing the amount of computation required.

There are two types of pooling: max pooling and average pooling. The difference between max pooling and average pooling lies in the extraction of the maximum and average values from the pooling area, respectively.
