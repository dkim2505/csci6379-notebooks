# Multiple Linear Regression

CSCI 6379 · Topic 8. Several input features, one weight each, same gradient descent. The data was built from w = (0.5, 1.0, 2.0), b = 0, so training should recover exactly those.


```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([[3.9, 2.1, 7.5, 19.05], [2.1, 3.2, 4.5, 13.25],
                 [4.7, 4.1, 7.8, 22.05], [8.5, 5.4, 1.2, 12.05],
                 [1.9, 3.4, 6.2, 16.75], [6.3, 4.3, 7.7, 22.85],
                 [8.9, 4.4, 3.3, 15.45]])
x, y = data[:, :-1], data[:, -1]
n = len(data)
print("x shape:", x.shape, "| y shape:", y.shape)
```

## One update line per weight

```python
w1 = w2 = w3 = b = 0.0
alpha = 0.01
for _ in range(50000):
    err = w1*x[:, 0] + w2*x[:, 1] + w3*x[:, 2] + b - y
    w1 -= alpha * (1/n) * np.sum(err * x[:, 0])
    w2 -= alpha * (1/n) * np.sum(err * x[:, 1])
    w3 -= alpha * (1/n) * np.sum(err * x[:, 2])
    b  -= alpha * (1/n) * np.sum(err)
print("w1=%.3f w2=%.3f w3=%.3f b=%.3f" % (w1, w2, w3, b))   # -> 0.5 1.0 2.0 0.0
```

## Much better: the vectorized version

One matrix product updates every weight at once, no matter how many features.

```python
w = np.zeros(3)
b = 0.0
alpha = 0.01
costs = []
for _ in range(50000):
    err = x @ w + b - y             # all residuals
    costs.append(np.mean(err**2))
    w = w - alpha * (1/n) * (x.T @ err)
    b = b - alpha * (1/n) * np.sum(err)
print("w =", np.round(w, 3), "b =", round(b, 3))
```

## The learning curve and the fit

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(costs); ax[0].set_yscale("log")
ax[0].set_xlabel("iteration"); ax[0].set_ylabel("cost (log)"); ax[0].set_title("cost -> 0")
pred = x @ w + b
ax[1].plot([10, 24], [10, 24], "--", color="gray")
ax[1].scatter(y, pred, c="red", s=60)
ax[1].set_xlabel("actual y"); ax[1].set_ylabel("predicted"); ax[1].set_title("perfect fit")
plt.tight_layout(); plt.show()
```
