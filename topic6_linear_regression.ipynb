# Linear Regression and MSE

CSCI 6379 · Topic 6. Predict an exam score from study hours, and measure how good a line is with Mean Squared Error.


```python
import numpy as np
import matplotlib.pyplot as plt

# four students: [study hours, exam score]
data = np.array([[2, 81], [4, 93], [6, 91], [8, 97]])
train_x = data[:, 0]     # study hours
train_y = data[:, 1]     # exam scores
print("x =", train_x)
print("y =", train_y)
```

## A line as a model:  H(x) = a x + b

Pick a guess for the slope `a` and intercept `b`, then predict every student at once.

```python
a, b = 3.4, 77
pred = a * train_x + b            # H(x) for all four students
print(pred)                       # [ 83.8  90.6  97.4 104.2]
```

## Scoring the line with MSE

Residual = prediction - truth. Square them (so they can't cancel and big misses hurt more), then average.

```python
mse = np.sum((a * train_x + b - train_y) ** 2) / len(train_x)
print("MSE =", mse)               # 26.6
```

## A better line has a lower MSE

Try `a = 2.3, b = 79` and compare.

```python
for a, b in [(3.4, 77), (2.3, 79)]:
    mse = np.mean((a * train_x + b - train_y) ** 2)
    print(f"y = {a}x + {b}  ->  MSE = {mse:.1f}")
```

## Plot the data and both lines

```python
xs = np.array([0, 10])
plt.figure(figsize=(6, 4))
plt.scatter(train_x, train_y, c="red", s=80, label="data", zorder=5)
plt.plot(xs, 3.4 * xs + 77, "--", label="y = 3.4x + 77  (MSE 26.6)")
plt.plot(xs, 2.3 * xs + 79, "-",  label="y = 2.3x + 79  (MSE 8.3)")
plt.xlabel("study hours"); plt.ylabel("exam score")
plt.legend(); plt.grid(alpha=.3); plt.show()
```
