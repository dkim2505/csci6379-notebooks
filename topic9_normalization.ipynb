# Normalization

CSCI 6379 · Topic 9. Rescale features to a common range so gradient descent descends a round bowl instead of a thin ravine.


```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# two features on a similar scale (easy case)
X1 = [[1, 2], [2, 1], [3, 2], [4, 3]]

# three features on wildly different scales (the hard case)
X2 = [[1.5, 2864, 2.3], [2.6, 8372, 1.8], [1.2, 6453, 2.2], [2.3, 9587, 3.7],
      [1.9, 2332, 3.1], [3.7, 8574, 1.5], [2.1, 7665, 2.3], [1.4, 2428, 1.8],
      [3.7, 9476, 3.2], [1.5, 3422, 2.4]]
```

## Min-Max scaling: map every feature to [0, 1]

`fit` learns each column's min and max; `transform` applies them.

```python
scaler = MinMaxScaler()
scaler.fit(X1)
print(scaler.transform(X1))
# [[0.    0.5 ]
#  [0.333 0.  ]
#  [0.667 0.5 ]
#  [1.    1.  ]]
```

```python
scaler2 = MinMaxScaler()
print(scaler2.fit_transform(X2).round(2))   # every column now in [0, 1]
```

## Standardization: mean 0, standard deviation 1

```python
print(StandardScaler().fit_transform([[1], [2], [3], [4]]).ravel().round(3))
# [-1.342 -0.447  0.447  1.342]
```

## The formulas, by hand

Min-Max: (x - min) / (max - min).  Standard score: (x - mean) / std.

```python
import numpy as np
x = np.array([1, 2, 3, 4.])
print("min-max :", ((x - x.min()) / (x.max() - x.min())).round(3))
print("z-score :", ((x - x.mean()) / x.std()).round(3))
```
