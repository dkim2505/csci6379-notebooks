# Gradient Descent

CSCI 6379 · Topic 7. Find the best line automatically by rolling downhill on the cost (MSE), instead of guessing.


```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([[2, 81], [4, 93], [6, 91], [8, 97]])
x, y = data[:, 0], data[:, 1]
n = len(data)
```

## The gradient-descent loop

Start from a bad line (w = b = -1), then repeatedly step each parameter opposite its gradient.

```python
w, b = -1.0, -1.0
alpha = 0.05                 # learning rate
costs = []

for _ in range(2000):
    costs.append(np.mean((w * x + b - y) ** 2))
    gw = (1/n) * np.sum((w * x + b - y) * x)   # slope in the w direction
    gb = (1/n) * np.sum( w * x + b - y)         # slope in the b direction
    w = w - alpha * gw
    b = b - alpha * gb

print("w = %.4f, b = %.4f, final cost = %.4f" % (w, b, costs[-1]))
# -> w = 2.3000, b = 79.0000, final cost = 8.3000  (the line we hand-picked last topic!)
```

## The learning curve

The cost falls from thousands to its minimum of 8.3.

```python
plt.figure(figsize=(6, 4))
plt.plot(costs)
plt.yscale("log")
plt.xlabel("iteration"); plt.ylabel("cost (MSE, log scale)")
plt.title("Cost drops as the line improves"); plt.grid(alpha=.3, which="both")
plt.show()
```

## The fitted line

```python
xl = np.linspace(0, 10, 100)
plt.figure(figsize=(6, 4))
plt.scatter(x, y, c="red", s=80, zorder=5)
plt.plot(xl, w * xl + b, c="green", label=f"y = {w:.2f}x + {b:.1f}")
plt.xlabel("study hours"); plt.ylabel("exam score")
plt.legend(); plt.grid(alpha=.3); plt.show()
```

## Try it: change the learning rate

Set `alpha = 0.07` above and re-run: the cost grows instead of shrinking (it diverges). Set `alpha = 0.01`: it converges, but much more slowly.
